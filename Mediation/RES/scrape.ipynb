```python
import pandas as pd
```

```python
with open("RES Mediation.txt", 'r') as file:
    content = file.read()
    lines = content.split('\n')
        
idx = 0
key_str = "The Review of Economic Studies"
titles = list()
doi_links = list()
dates = list()

while idx < len(lines):
    line = lines[idx]
    if key_str not in line:
        idx += 1
        continue
    title = lines[idx - 2]
    date_line = lines[idx + 1]
    date_line_words = date_line.split()
    date = date_line_words[-1]
    
    if title.endswith("Get access Arrow"):
        title = title.removesuffix("Get access Arrow")
    if title.endswith("Open Access"):
        title = title.removesuffix("Open Access")

    
    titles.append(title)
    idx += 1
    start = line.find("https://doi.org/")
    
    end = line.find(" ", start)
    if end == -1:
        doi_link = line[start:]
    else:
        doi_link = line[start:end]
    doi_links.append(doi_link)
    
    dates.append(date)

df = pd.DataFrame({
    "Title": titles,
    "Link": doi_links,
    "Year": dates
})
```

```python
# see if all the titles look right
for title in titles:
    print(title)
```

```python
df["Journal"] = "RES"
df["Category"] = "Mediation"

new_order = ["Journal", "Title", "Link", "Category", "Year"]
df = df[new_order]
```
