```python
import pandas as pd
```

```python
with open("QJE Surrogates.txt", 'r') as file:
    content = file.read()
    lines = content.split('\n')
        
idx = 0
key_str = "The Quarterly Journal of Economics"
titles = list()
doi_links = list()

while idx < len(lines):
    line = lines[idx]
    if key_str not in line:
        idx += 1
        continue
    title = lines[idx - 2]
    
    if title.endswith("Get access Arrow"):
        title = title.removesuffix("Get access Arrow")
    if title.endswith("Open Access"):
        title = title.removesuffix("Open Access ")

    
    titles.append(title)
    idx += 1
    start = line.find("https://doi.org/")
    
    end = line.find(" ", start)
    if end == -1:
        doi_link = line[start:]
    else:
        doi_link = line[start:end]
    doi_links.append(doi_link)

df = pd.DataFrame({
    "Title": titles,
    "Link": doi_links
})
```

```python
# see if all the titles look right
for title in titles:
    print(title)
```

```python
# see if all the links look right
for link in doi_links:
    print(link)
```

```python
print(len(titles))
print(len(doi_links))

# Did we get the right number of results
```

```python
df["Journal"] = "QJE"
df["Category"] = "Surrogates"

new_order = ["Journal", "Title", "Link", "Category"]
df = df[new_order]
```

```python
df.to_csv("qje_surrogates.csv", index=False)
```
